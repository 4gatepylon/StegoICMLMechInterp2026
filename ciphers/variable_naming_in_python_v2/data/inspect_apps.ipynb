# Inspect filtered APPS problems

Load [APPS on Hugging Face](https://huggingface.co/datasets/codeparrot/apps) using the v2 loader. This notebook **only prints data**: it never executes reference solutions or supplied cases. “Ground truth” means an answer supplied by APPS, not correctness independently established here.

Use the `stego` environment with the [CPU or GPU requirements](../../../README.md#setup) installed. Set `STEGO_ARTIFACTS_DIR` to your artifact directory before starting the kernel; downloads go into `datasets/apps` below it. Relative artifact directories resolve from the repository root. The loader requires `datasets>=5.0.1` for heterogeneous JSON test data.


```python
import json
import os
import sys
from pathlib import Path

# Jupyter may start in this folder; repository imports must start at the repo root.
repo_root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "ciphers/variable_naming_in_python_v2/data/apps.py").is_file())
os.chdir(repo_root)
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
```

```python
from ciphers.variable_naming_in_python_v2.data.apps import AppsConfig, load_apps
```

The easy tier is `introductory`; other choices are `interview` and `competition`. All bounds are inclusive. Line counts include blank/comment lines; character counts include whitespace. A problem remains if **any reference solution qualifies**, and only qualifying solutions are returned. `min_tests` counts supplied input/output pairs, not assertions or coverage.


```python
config = AppsConfig(
    split="train",
    difficulties=("introductory",),
    min_lines=20,
    max_lines=None,
    min_chars=0,
    max_chars=None,
    min_tests=10,
)

preview_problems = 3
preview_solutions = 2
preview_tests = 3
sample_seed = 42

problems = load_apps(config)
print(config.model_dump_json(indent=2))
print(f"Retained {len(problems):,} problems")
if not len(problems):
    print("No matches. Adjust the difficulty, answer-size bounds, or min_tests above, then rerun.")
```

Each row supplies `question`, filtered `solutions`, parsed `input_output`, `num_tests`, and source metadata (`problem_id`, `difficulty`, `url`, `starter_code`). In `input_output`, `inputs[i]` pairs with `outputs[i]`; `fn_name=None` denotes standard-input tasks and a string denotes function-call tasks. Values are preserved as JSON, including nested arguments. The preview prints a reproducible sample and a configurable number of answers and cases.


```python
sample = problems.shuffle(seed=sample_seed).select(range(min(preview_problems, len(problems))))
for problem in sample:
    print("=" * 88)
    print(f"APPS {config.split}/{problem['problem_id']} | {problem['difficulty']} | {problem['num_tests']} supplied cases")
    print(f"Source: {problem['url']}")
    print("\nQUESTION\n")
    print(problem["question"])
    if problem["starter_code"]:
        print("\nSTARTER CODE\n")
        print(problem["starter_code"])

    print(f"\nQUALIFYING GROUND-TRUTH ANSWERS ({len(problem['solutions'])} total)")
    for index, answer in enumerate(problem["solutions"][:preview_solutions], start=1):
        print(f"\nAnswer {index}: {len(answer.splitlines())} lines, {len(answer)} characters\n")
        print(answer)

    cases = problem["input_output"]
    print(f"\nSUPPLIED CASES | entry point: {cases['fn_name'] or 'standard input/output'}")
    for index, (case_input, expected_output) in enumerate(zip(cases["inputs"][:preview_tests], cases["outputs"][:preview_tests], strict=True), start=1):
        print(f"\nCase {index}")
        print("Input:", json.dumps(case_input, ensure_ascii=False, indent=2))
        print("Expected output:", json.dumps(expected_output, ensure_ascii=False, indent=2))
```
